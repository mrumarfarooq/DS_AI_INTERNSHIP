Machine Learning (ML) is a way of training computers to learn patterns from data rather than programming them with strict, step-by-step rules. In traditional programming, we feed the data to the computer and then program it with all the steps the computer must take in order to arrive at the correct output. The computer then follows these rules to the letter. But in Machine Learning, the process is not the same. Rather than programming the computer to solve a problem, we provide the computer with a lot of data and the correct solutions
As time passes and the computer is fed more data, it becomes smarter and better at making predictions. Eventually, after the computer has been trained on a lot of data, it can then use this knowledge to make informed decisions about new, unseen data.

TYPES OF MACHINE LEARNING:-

️ "Supervised Learning"

Supervised Learning is a type of Machine Learning where the data already contains the correct answers, also called labels.
The model learns by comparing its predictions with the actual answers and adjusting itself to reduce errors. It is like learning with a teacher 
who provides the correct solutions.

️ "Unsupervised Learning"

Unsupervised Learning is used when the data does not have labeled answers. The model tries to explore the data on its own and discover hidden patterns,
structures, or clusters. There is no “teacher” guiding it with correct outputs. Instead, the algorithm finds similarities and differences within the data.

 "Reinforcement Learning"

Reinforcement Learning is a learning method where an agent learns by interacting with an environment.
Instead of being given correct answers, it receives rewards or penalties based on its actions. Over time, it learns which actions maximize rewards and 
minimize penalties.

Using the wrong kind of learning is one of the most common pitfalls for a beginner in Machine Learning. If your data is labeled and you are doing
unsupervised learning, your model will not be able to make use of the available data. If your data is not labeled and you are doing supervised learning,
it will not work at all.
